## 1. AI System Risk Score

```python
impact_score = 8
likelihood_score = 6
# Calculate a simple risk score:
risk_score = impact_score * likelihood_score
print(f"Risk_Score: {risk_score: .2f}")
```

## 2. Governance Control Coverage

```python
# Suppose an AI system has
total_controls = 20
implemented_controls = 17
#Calculate control coverage: 
control_coverage = ( implemented_controls / total_controls )* 100
print(f"Control Coverage: {control_coverage:.2f}%")
```

## 3. Approval Completion

```python
total_requirements = 10
approved_requirements = 9
# Calculate the approval completion percentage:
approval_completion = ( approved_requirements / total_requirements)*100
print(f"Approval Completion: {approval_completion:.2f}%")
```
