# Exercise 02 – Working with Types
## AI Governance Engineering Application

This experiment applies Python data types to AI governance,
risk assessment and control monitoring scenarios.

```python
# AI Governance - Python Data Types

system_name = "AI Quality Assistant"
risk_score = 75.5
control_count = 12
governance_status = "Under Review"
approval_required = True

print(system_name)
print(risk_score)
print(control_count)
print(governance_status)
print(approval_required)
```

```python
print(type(system_name))
print(type(risk_score))
print(type(control_count))
print(type(governance_status))
print((approval_required))
```

```python
risk_score_condition = bool(approval_required)
print(risk_score_condition)
print(type(risk_score_condition))
```

```python
control_count_decimal = float(control_count)
print(control_count_decimal)
print(type(control_count_decimal))
```
